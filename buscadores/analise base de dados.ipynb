```python
import pandas as pd
import re
```

```python
df = pd.read_csv('./dados_conjuntos.csv')
```

```python
padrao = r"""[^\w\s\.\-_:\/áéíóúâêîôûàèìòùãõçÁÉÍÓÚÂÊÎÔÛÀÈÌÒÙÃÕÇ]"""
```

```python
def limpar_coluna(df, nome_coluna):
    """
    Remove caracteres especiais da coluna especificada de um DataFrame,
    preservando letras acentuadas, números, espaços e os seguintes sinais de pontuação:
    hífen (-), underscore (_), ponto (.), dois pontos (:) e barra (/)
    
    Parâmetros:
    - df: pandas.DataFrame
    - nome_coluna: str
    
    Retorna:
    - DataFrame com a coluna limpa
    """
    
    def limpar_texto(texto):
        texto = str(texto)
        texto = re.sub(r'[\t\n\r\f\v]', ' ', texto)        # substitui tabulações e quebras de linha por espaço
        texto = re.sub(padrao, '', texto)                  # remove caracteres indesejados
        texto = re.sub(r'\s+', ' ', texto).strip()         # remove múltiplos espaços e espaços nas bordas
        return texto
    
    df[nome_coluna] = df[nome_coluna].apply(limpar_texto)
    return df
```

```python
df = limpar_coluna(df, 'nomeRecurso')
```

```python
# selecionar apenas dados com periodicidade e com formatos de recuros validos
df = df[(df.formato.str.contains(r'csv|cvs|xls|zip|json|xml|rar|html', case=False, na=True))]
# df = df[(df.periodicidade != 'SOB_DEMANDA') & (df.periodicidade.notna()) & (df.formato.str.contains(r'csv|cvs|xls|zip|json|xml|rar|html', case=False, na=True))]
```

```python
df_varios_dataset = df[['id','titulo','formato']].groupby(['id','formato']).count().reset_index()
df_varios_dataset = df_varios_dataset[df_varios_dataset.titulo > 2]
```

```python
df = df[df.id.isin(df_varios_dataset.id)]
```

```python
# removendo linhas duplicadas
df = df[['id', 'titulo']].drop_duplicates()
```
