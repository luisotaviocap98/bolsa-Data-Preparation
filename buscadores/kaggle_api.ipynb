```python
import kaggle
import time
import pandas as pd
```

```python
k = kaggle.api
```

```python
k.authenticate()
```

```python
datasets = []
```

```python
p=1
while True:
    try:
        d = k.dataset_list(file_type='csv', license_name='odb', page= p)
        if not d:
            print('Não achou pagina',p)
            break
        datasets.extend(d)
        p+= 1
        time.sleep(1)  # Aguardar 1 segundo entre as requisições para evitar sobrecarga
    except Exception as e:
        print('Erro na pagina', p, e)
        break
```

```python
len(datasets)
```

```python
df = pd.DataFrame([d.to_dict() for d in datasets])
```

```python
df.to_csv('./datasets_kaggle.csv', index=False)
```
