```python
import pandas as pd
```

```python
df = pd.read_csv('./dados_worldbank.csv')
```

```python
df
```

```python
# selecionar apenas dados com periodicidade e com formatos de recuros validos
df = df[(df.formato_fonte.str.contains(r'csv|cvs|xls|zip|json|xml|rar|api|excel|html', case=False, na=True)) & (df.periodicidade.notna())]
```

```python
df
```

```python
# remover linhas duplicadas
df = df[['dataset_id', 'dataset_unique_id', 'nome_dataset', 'link', 'link_antigo']].drop_duplicates()
```

```python
df.to_csv('./candidatos_schema_matching_worldbank.csv', index=False)
```
