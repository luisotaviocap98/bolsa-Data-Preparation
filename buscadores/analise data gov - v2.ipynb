```python
import pandas as pd
from fuzzywuzzy import fuzz
import numpy as np
```

```python
df = pd.read_csv('./datasets_datagov_eua.csv',sep='|')
```

```python
df
```

```python
df.formato_arquivo = df.formato_arquivo.str.upper()
```

```python
df_valido = df[(df.formato_arquivo.str.contains(r'csv|cvs|xls|zip|json|rar|comma|tsv|excel', case=False, na=False))]
df_invalido = df[(df.formato_arquivo.str.contains(r'geo|xml|pdf', case=False, na=False))]
# df = df[(df.formato_arquivo.str.contains(r'csv|cvs|xls|zip|json|xml|rar|comma|tsv|excel', case=False, na=True))]
```

```python
df_valido = df_valido[(df_valido.qtd_arquivos > 0)]
df_invalido = df_invalido[(df_invalido.qtd_arquivos > 0)]
# df = df[(df.qtd_arquivos > 1) | (df.qtd_arquivos.isna())]
```

```python
df = df[df.dataset_id.isin(df_valido.dataset_id)]
df = df[~df.dataset_id.isin(df_invalido.dataset_id)]
```

```python
df = df[(df.nivel_acesso == 'public') | (df.nivel_acesso.isna())]
# df = df[(df.nivel_acesso == 'public') | (df.nivel_acesso.isna())]
```

```python
df
```

```python
# Buscar por strings parecidas

df_similar = df[['dataset_id','nome','titulo']].drop_duplicates()

df_similar['titulo_tratado'] = df_similar['titulo'].str.replace(r'[^a-zA-Z\s]', '', regex=True).str.replace(r'\s+', ' ', regex=True).str.strip()

# Criar chave com primeiras 3 palavras
# df_similar['chave'] = df_similar['titulo'].str.lower().str.split().str[1:-2].str.join(' ')
# df_similar['chave'] = np.where(
#     df_similar['titulo_tratado'].str.split().str.len() <= 4,
#     df_similar['titulo_tratado'].str.lower(),
#     df_similar['titulo_tratado'].str.lower().str.split().str[1:-2].str.join(' ')
# )

# Contar ocorrências
contagem = df_similar['titulo_tratado'].value_counts()

# Marcar duplicatas
df_similar['tem_similar'] = df_similar['titulo_tratado'].map(contagem) > 1

# df_similar['tem_similar'] = df_similar['titulo'].apply(lambda x: tem_similar(x, df_similar))
df_similar = df_similar[df_similar['tem_similar']]
```

```python
# Datasets com mais de um arquivo do mesmo tipo

df_qnt = df.groupby(['dataset_id','formato_arquivo'])['id_arquivo'].count().to_frame().reset_index()
df_qnt = df_qnt[df_qnt.id_arquivo > 1]
```

```python
df = df[df.dataset_id.isin(df_qnt.dataset_id) | df.dataset_id.isin(df_similar.dataset_id)]
```

```python
df[['dataset_id','qtd_arquivos','url','titulo']].drop_duplicates().sort_values(by=['titulo']).to_excel('./dataset_candidatos_datagov_eua.xlsx', index=False)
# df.drop_duplicates().to_excel('./dataset_candidatos_datagov_eua.xlsx', index=False)
```
