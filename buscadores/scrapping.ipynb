```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
import time

# Configurações do Chrome headless
options = Options()
options.add_argument('--headless')
options.add_argument('--disable-gpu')
options.add_argument('--no-sandbox')

driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=options)

try:
    url = "https://www.kaggle.com/datasets/wasiqaliyasir/breast-cancer-dataset"
    driver.get(url)

    # Espera o carregamento (pode ser melhor com WebDriverWait)
    time.sleep(8)

    imgs = driver.find_elements(By.TAG_NAME, "img")
    resultados = []

    for img in imgs:
        src = img.get_attribute("src")
        if src and src.startswith("https://www.kaggle.com/static/images/medals/datasets/"):
            resultados.append(src)

    if resultados:
        print("Imagens encontradas:")
        for src in resultados:
            print(src)
    else:
        print("Nenhuma imagem com esse prefixo foi encontrada.")

finally:
    driver.quit()
```
