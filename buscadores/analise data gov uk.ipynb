```python
import pandas as pd
from urllib.parse import urlparse
import requests

df = pd.read_csv('./datasets_datagov_uk.csv',sep='|')
```

```python
df = df[df.dataset_aberto == True]
```

```python
df_varios_dataset = df[['dataset_nome','recurso_id']].groupby('dataset_nome')['recurso_id'].count().reset_index()
df_varios_dataset = df_varios_dataset[df_varios_dataset.recurso_id == 1]
df_interesse = df[df.dataset_nome.isin(df_varios_dataset.dataset_nome)]
```

```python
df_interesse = df_interesse[df_interesse.recurso_url.notna()]
df_interesse = df_interesse[df_interesse.recurso_url != '']
```

```python
df_valido = df_interesse[(df_interesse.recurso_formato.str.contains(r'csv|cvs|xls|zip|rar|comma|tsv|excel|gz', case=False, na=False))]
```
