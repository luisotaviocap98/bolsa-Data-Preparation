```python
import requests
import pandas as pd
import time
```

```python
headers = {
    'accept': 'application/json'
}
```

```python
url_datasets = 'https://catalog.data.gov/api/3/action/package_list?rows=1000'
```

```python
lista_dataset =[]

for i in range(0, 312):
    url_lista = f'{url_datasets}&start={i*1000}'
    resposta_lista = requests.get(url_lista, headers=headers)
    
    if resposta_lista.status_code != 200:
        continue
    
    registros = resposta_lista.json()
    
    if not registros or not registros.get('result') or not registros.get('result').get('results'):
        continue
    
    
    for r in registros.get('result').get('results'):
        if not r.get('resources'):
            continue
        
        for f in r.get('resources'):
            dicionario = {
                'dataset_id':r.get('id',''),
                'qtd_arquivos':r.get('num_resources',''),
                'privado':r.get('private',''),
                'tipo':r.get('type',''),
                'nome':r.get('name',''),
                'titulo':r.get('title',''),
                'descricao':r.get('notes',''),
                'url':r.get('url',''),
                'nivel_acesso':next((item["value"] for item in r.get('extras') if item["key"] == "accessLevel"), None),
                'formato_arquivo':f.get('format',''),
                'id_arquivo':f.get('id',''),
                'url_arquivo':f.get('url','')
            }
            
            lista_dataset.append(dicionario)
        
    time.sleep(0.1)
```

```python
df = pd.DataFrame(lista_dataset)
df.to_csv('C:/Users/user/Desktop/bolsa estudo/datasets_datagov_eua.csv', index=False, encoding='utf-8',sep='|')
```
