```python
import pandas as pd
from bs4 import BeautifulSoup
from googletrans import Translator
```

```python
# Caminho do arquivo original
input_file = 'lista_dataset_datagov_eua.xlsx'
output_file = 'lista_dataset_datagov_eua_traduzido.xlsx'

# Inicializar tradutor
translator = Translator()

def limpar_html(texto_html):
    """Remove tags HTML e faz limpeza de espaços."""
    if pd.isna(texto_html):
        return ""
    texto_limpo = BeautifulSoup(str(texto_html), "html.parser").get_text(separator=' ')
    texto_limpo = texto_limpo.replace('\n', ' ').replace('\r', ' ')
    texto_limpo = ' '.join(texto_limpo.split())  # Remove espaços e quebras duplicadas
    return texto_limpo.strip()

def traduzir_texto(texto):
    """Traduz o texto para português."""
    if not texto.strip():
        return ""
    try:
        traducao = translator.translate(texto, src='en', dest='pt').text
        return traducao.strip()
    except Exception as e:
        return f"Erro de tradução: {e}"

# Ler o Excel original
df = pd.read_excel(input_file)

# Limpar HTML
df['descricao_limpa'] = df['descricao'].apply(limpar_html)

# Traduzir linha a linha
df['descricao_traduzida'] = df['descricao_limpa'].apply(traduzir_texto)

# Salvar o novo Excel
df.to_excel(output_file, index=False)
```
